
# Agentic AI for Anomaly Detection (Class-Based Teaching Notebook)

This notebook implements your full **agentic AI architecture using classes**, aligned with:

1. Step 1 — Model
2. Step 2 — Threshold
3. Step 3 — Profiling

Each step is implemented as an **Agent Class**, so students clearly see the architecture.


```

# Install if needed
# !pip install pyod

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 0 — Data Generation

```

from pyod.utils.data import generate_data

X_train, X_test, y_train, y_test = generate_data(
    n_train=500,
    n_test=500,
    n_features=6,
    contamination=0.05,
    random_state=123
)

X_train = 5 - X_train
X_test = 5 - X_test
```

## Shared Model Factory

```

from pyod.models.hbos import HBOS
from pyod.models.ecod import ECOD
from pyod.models.iforest import IForest
from pyod.models.gmm import GMM

class ModelFactory:
    def create(self, name):
        if name == 'HBOS':
            return HBOS(n_bins=50, contamination=0.05)
        elif name == 'ECOD':
            return ECOD()
        elif name == 'IForest':
            return IForest(contamination=0.05)
        elif name == 'GMM':
            return GMM(contamination=0.05)
        else:
            raise ValueError("Unknown model")
```

## Agent 1 — Model Agent (Step 1)

```

class ModelAgent:

    def __init__(self, model_name):
        self.model_name = model_name
        self.factory = ModelFactory()

    def run(self, X_train, X_test):
        model = self.factory.create(self.model_name)
        model.fit(X_train)

        return {
            "model": model,
            "train_scores": model.decision_function(X_train),
            "test_scores": model.decision_function(X_test),
            "threshold": getattr(model, 'threshold_', None)
        }

# Example
model_agent = ModelAgent("HBOS")
model_result = model_agent.run(X_train, X_test)
scores = model_result['train_scores']
```

## Agent 2 — Threshold Agent (Step 2)

```

class ThresholdAgent:

    def run(self, scores):
        threshold = np.percentile(scores, 95)
        return threshold

threshold_agent = ThresholdAgent()
threshold = threshold_agent.run(scores)

plt.hist(scores, bins='auto')
plt.title("Score Distribution")
plt.show()

threshold
```

## Agent 3 — Profiling Agent (Step 3)

```

class ProfilingAgent:

    def run(self, X, scores, threshold):
        df = pd.DataFrame(X)
        df['score'] = scores
        df['group'] = np.where(scores < threshold, 'Normal', 'Outlier')

        count = df.groupby('group').size().reset_index(name='count')
        count['perc'] = count['count'] / count['count'].sum() * 100

        stats = df.groupby('group').mean().reset_index()

        return count.merge(stats, on='group')

profiling_agent = ProfilingAgent()
profile = profiling_agent.run(X_train, scores, threshold)
profile
```

## Agent 4 — Ensemble Agent

```

from pyod.models.combination import average
from pyod.utils.utility import standardizer

class EnsembleAgent:

    def run(self, X_train, model_list):
        scores = []

        for m in model_list:
            model = ModelFactory().create(m)
            model.fit(X_train)
            scores.append(model.decision_function(X_train))

        scores = np.array(scores).T
        norm_scores, _ = standardizer(scores, scores)

        return average(norm_scores)

ensemble_agent = EnsembleAgent()
ensemble_scores = ensemble_agent.run(X_train, ['HBOS','ECOD','IForest'])

plt.hist(ensemble_scores, bins='auto')
plt.title("Ensemble Scores")
plt.show()
```

## Agent 5 — Explainability Agent (ECOD)

```

class ExplainabilityAgent:

    def run(self, X_train):
        model = ECOD()
        model.fit(X_train)

        scores = model.decision_function(X_train)
        idx = np.where(scores > np.percentile(scores, 95))[0][:3]

        for i in idx:
            model.explain_outlier(i)

ExplainabilityAgent().run(X_train)
```

## Agent 6 — Triangulation Agent

```

class TriangulationAgent:

    def run(self, X_train, X_test, models):
        preds = []

        for m in models:
            model = ModelFactory().create(m)
            model.fit(X_train)
            preds.append(model.predict(X_test))

        preds = np.array(preds)
        final = np.sum(preds, axis=0) >= 2

        return final

tri_agent = TriangulationAgent()
final_pred = tri_agent.run(X_train, X_test, ['HBOS','ECOD','IForest'])

print('Consensus outliers:', np.sum(final_pred))
```

## Full Pipeline (Agent Orchestration)

```

class AnomalyPipeline:

    def __init__(self, model_name):
        self.model_agent = ModelAgent(model_name)
        self.threshold_agent = ThresholdAgent()
        self.profiling_agent = ProfilingAgent()

    def run(self, X_train, X_test):
        result = self.model_agent.run(X_train, X_test)

        threshold = self.threshold_agent.run(result['train_scores'])

        profile = self.profiling_agent.run(
            X_train,
            result['train_scores'],
            threshold
        )

        return profile

pipeline = AnomalyPipeline('HBOS')
pipeline.run(X_train, X_test)
```
